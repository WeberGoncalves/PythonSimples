```python
# Conversão de string para número
try:
    numero = int("abc")  # erro: não é número
    print(numero)
except ValueError:
    print("Erro: não foi possível converter para inteiro.")
```

```python
# Divisão por zero
try:
    resultado = 10 / 0
    print(resultado)
except ZeroDivisionError:
    print("Erro: divisão por zero não é permitida.")
```

```python
#Acesso a chave inexistente em dicionário
aluno = {"nome": "Maria", "nota1": 8}
try:
    print(aluno["nota2"])
except KeyError:
    print("Erro: chave 'nota2' não encontrada no dicionário.")
```

```python
#Capturar múltiplos tipos de erro
try:
    valor = int(input("Digite um número: "))
    resultado = 10 / valor
    print("Resultado:", resultado)
except ValueError:
    print("Erro: você não digitou um número válido.")
except ZeroDivisionError:
    print("Erro: não é possível dividir por zero.")
```
